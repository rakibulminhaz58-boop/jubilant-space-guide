```python
import os
from dotenv import load_dotenv

load_dotenv()
```

### GENERATIVE AI APPLICATION USING GROQ AI MODEL

```python
from langchain.chat_models import init_chat_model
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model(
    model="openai/gpt-oss-20b",
    model_provider="groq",
)
```

### FIRST GENERATIVE AI APPLICATION

```python
response = model.invoke("Write an essay on AI")
print(response.content)
```

### GENERATIVE AI WITH GEMINI API KEY

```python
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
```

```python
from langchain.chat_models import init_chat_model

model = init_chat_model(
    model="gemini-3.5-flash-lite",
    model_provider="google_genai"
)

response = model.invoke("hello ai")
print(response.content)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
model = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
```

```python
response = model.invoke("Write an essay on AI")
print(response.content)
```

### STREAMING AND BATCH

```python
##NORMAL WAY
model.invoke("write me an essay with 1000 words")
```

```python
## Streaming way
for chunk in model.stream("write me an essay with 1000 words"):
    print(chunk.text, end="", flush=True)
```

### BATCH

```python
responses=model.batch(
    ["how are you?",
    "what you do?",
    'what you eat?']
)
for response in responses:
    print(response)
```
